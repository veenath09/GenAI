```python
import tensorflow as tf
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('diabetes.csv')
data
```

```python
print(type(data))
print(data.dtypes)
```

```python
data.sum()
```

```python
data.describe()
```

```python
scaler = StandardScaler()
```

```python
y = data['Outcome']
x = data.drop(columns=['Outcome'])
```

```python
y
```

```python
x
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.1 , random_state = 45)
```

```python
scaler = StandardScaler()
x_train = pd.DataFrame(scaler.fit_transform(x_train))
x_test = pd.DataFrame(scaler.transform(x_test))
```

```python
x_test
```

```python
x_train.shape[1]
```

```python
model = keras.Sequential()
```

```python
model = keras.Sequential([
 keras.layers.Input(shape=(x_train.shape[1],)),
 keras.layers.Dense(5, activation="relu"),
 keras.layers.Dropout(0.3), # 30% neurons dropped
 keras.layers.Dense(3, activation="relu"),
 keras.layers.Dropout(0.2), # 20% neurons dropped
 keras.layers.Dense(1, activation="sigmoid")
])
```

```python
optimzer = keras.optimizers.Adam



model.compile(optimizer = optimzer(learning_rate=0.001),loss = keras.losses.BinaryCrossentropy(),metrics = ['accuracy'])


early_stopping_callback = keras.callbacks.EarlyStopping(
    monitor='val_loss',  # Monitor validation loss
    patience=5,          # Stop after 5 epochs of no improvement
    min_delta=0.001,     # Minimum change to qualify as improvement
    restore_best_weights=True # Restore weights from the best epoch

)
```

```python
model.fit(x_train,y_train,batch_size=16,epochs=150,verbose=True,validation_split=0.2)
```

```python
from sklearn.metrics import f1_score
```

```python
y_pred = (model.predict(x_test)>0.5).astype(int)
```

```python
y_pred = [1 if val > 0.5 else 0 for val in y_pred]
y_pred
```

```python
y_pred1 = y_pred.reshape(-1,)
y_pred1.shape
```

```python
f1 = f1_score(y_test,y_pred)
```

```python
f1
```

```python
print(f"Test F1 Score: {f1:.4f}")
```

```python
from sklearn.metrics import confusion_matrix , precision_score , recall_score

cm = confusion_matrix(y_test,y_pred)
cm
```

```python
import seaborn as sns

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Pred 0','Pred 1'], yticklabels=['Actual 0','Actual 1'])
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.title('Confusion Matrix')
plt.show()
```

```python
history = model.fit(x_train, y_train, epochs=150, validation_data=(x_train, y_train))

# Access the history dictionary
print(history.history.keys())

# Plotting the training history
plt.figure(figsize=(12, 6))

# Plot loss
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
if 'val_loss' in history.history:
    plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

# Plot accuracy
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Training Accuracy')
if 'val_accuracy' in history.history:
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
plt.show()
```
